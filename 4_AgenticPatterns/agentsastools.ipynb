# Learning how to create tools with OPENAISDK


```python
from agents import Agent, Runner, trace
from dotenv import load_dotenv
from typing import Any

load_dotenv()
```

# Creating First Agent
- the system will go with default set model for agent model settings
- can see the default model on traces
refer to documentation: https://openai.github.io/openai-agents-python/models/

```python
agent1 = Agent[Any](
    name="assistant",
    model="gpt-6-luna",
    instructions="You are a helpful assistant"
)
```

```python
result1 = await Runner.run(
    agent1,
    "Write a 3-sentence summary of how to best learn a new technology."
)
```

```python

print(result1)
print("\nLLM response for the prompt")
print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
print(result1.final_output)
```

# Running LLM with a trace

```python
result2 = await Runner.run(
    agent1,
    "What is the most important technology to learn for a software engineer?"
)
```

```python
with trace("Asking Advice"):
        result = await Runner.run(agent1, "What is the most important technology to learn for a software engineer?")
        print(result.final_output)
```

## model settings Verbosity
- low
- medium
- high

- low --> quick, simple answer
- medium --> balanced thining
- high --> more careful / detailed thinking
- max --> max
