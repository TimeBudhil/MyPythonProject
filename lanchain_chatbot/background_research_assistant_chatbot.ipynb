# Background Research Assistant
This chatbot is a research assistant that will host as many chats as you need. 
The use case of this assistant it to get overarching background research for a specific research question. 
#### Example: Postcolonial Research
For example, if you will want to look into Post-Colonial Research, you would receive the following report:

```
    A few quick questions to tailor this:
    - Are you focusing on a particular region (Africa, Caribbean, South Asia, Pacific, Latin America, diaspora) or period?
    - Is this for a course, a paper/thesis, or general reading?
    - Do you want primary literary texts, critical theory, recent scholarship, or all of the above?

    While you decide, here’s a concise primer and starting toolkit.

    What post‑colonial literature studies
    - Examines literature produced in or about societies shaped by European colonialism and its aftermath, and by diasporic communities.
    - Focuses on power, identity, language, resistance, nation-building, memory, hybridity, migration, and cultural representation.

    Core critical concepts and theorists (useful for foundations)
    - Orientalism — Edward Said (1978)
    - Subaltern studies and voice — Gayatri Chakravorty Spivak (“Can the Subaltern Speak?”, 1988)
    - Hybridity, mimicry, ambivalence — Homi K. Bhabha (The Location of Culture, 1994)
    - Colonial psychology and violence — Frantz Fanon (Black Skin, White Masks, 1952; The Wretched of the Earth, 1961)
    - Empire Writes Back — Ashcroft, Griffiths, Tiffin (The Empire Writes Back, 1989) — key overview of postcolonial literatures
    - Aimé Césaire (negritude) and other anti‑colonial movements

    Representative primary literary authors and texts (good starting list)
    - Chinua Achebe — Things Fall Apart (1958)
    - Ngũgĩ wa Thiong’o — Decolonising the Mind (1986) and novels (e.g., Petals of Blood)
    - Salman Rushdie — Midnight’s Children (1981)
    - Jean Rhys — Wide Sargasso Sea (1966)
    - Jamaica Kincaid — A Small Place (1988)
    - Derek Walcott — Omeros (1990)
    - Arundhati Roy — The God of Small Things (1997)
    - Chimamanda Ngozi Adichie — Half of a Yellow Sun (2006), Americanah (2013)
    - Tsitsi Dangarembga — Nervous Conditions (1988)
    - A. R. Ammons, Rabindranath Tagore, V. S. Naipaul (controversial but influential) — choose depending on region and theme

    Key secondary books and handbooks
    - The Empire Writes Back: Theory and Practice in Post‑Colonial Literatures — Ashcroft, Griffiths, Tiffin
    - Postcolonial Theory: A Critical Introduction — Leela Gandhi
    - The Cambridge Companion to Postcolonial Literary Studies — volumes on region and theory

    Useful journals and databases
    - Journals: Postcolonial Studies; Interventions; Journal of Postcolonial Writing; Small Axe (Caribbean); Research in African Literatures
    - Databases: JSTOR, Project MUSE, Google Scholar, ProQuest Dissertations/Theses
    - Archives & digital collections: British Library colonial collections, national libraries of former colonies; some oral-history projects

    Research strategies and methods
    - Narrow by region, period, and theme (e.g., language and translation; diaspora identity; gender and nationalism).
    - Balance primary texts with theory: pair a novel with one or two theoretical works that illuminate questions you care about.
    - Consider comparative approaches (e.g., two postcolonial novels from different regions) or close reading plus historical/contextual archival research.
    - Pay attention to language politics (writing in English vs native languages, translation, stylistic hybridity).
    - Include voices from the margins (women writers, indigenous authors, subaltern perspectives).

    Possible research questions / paper topics
    - How do postcolonial novels construct national identity after independence?
    - The role of language: authorial choice of colonizer’s language vs indigenous languages.
    - Representations of memory and trauma in postcolonial narratives.
    - Gender and postcolonial critique: intersections of patriarchy and colonial power.
    - Diasporic subjectivity and return narratives.

    How I can help next
    - Compile a tailored reading list (region or theme specific).
    - Build an annotated bibliography for a paper or lit review.
    - Find recent scholarly articles (I can search academic databases — I can run a targeted search for papers if you tell me keywords, region, or authors).
    - Suggest a thesis statement and outline for a research paper.
```
