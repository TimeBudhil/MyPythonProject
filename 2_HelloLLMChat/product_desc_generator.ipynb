# Product Description Generator. in class activity
The goal is to write a python function that generates a product description using dynamic inputs. 

### Requirements
- Create a function named generate_product_description()
- Pass dynamic inputs such as:
    - product name
    - target customer
- use client.responses.create() to geenrate content
- output should be short, attractive, customer-focused product description
- test function by calling it with different input values

### example use case
Generate description for a product like a "smart water bottle" for "college students'


```python
from dotenv import load_dotenv

import os

load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")
if openai_api_key:
    print("OPENAI_API_KEY is set")
else:
    print("OPENAI_API_KEY is not set")
```

```python
from openai import OpenAI
from pydantic import BaseModel

class ChatResponse(BaseModel):
    product_name: str
    target_consumer: str
    target_consumer_wants: str
    product_description: str
    product_unique_selling_point: str

system_prompt = "Answer eloquently (fewer, but more accurate words), and focus on customer-facing business value. i.e. a customer will want to be enticed by your product description."

def generate_product_description(message, history):
    client = OpenAI()
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}] 
    response = client.chat.completions.parse(model="gpt-6-luna", messages=messages, response_format=ChatResponse)
    result = response.choices[0].message.parsed
    return f"""
### {result.product_name}

**Target consumer:** {result.target_consumer}

**What they want:** {result.target_consumer_wants}

**Product description:**  
{result.product_description}

**Unique selling point:**  
{result.product_unique_selling_point}
"""
```

```python
import gradio as gr
gr.ChatInterface(generate_product_description, title="Conversation with LLM").launch(share=True, inbrowser=True,
inline=False)
```
