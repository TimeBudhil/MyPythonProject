```python
from dotenv import load_dotenv

import os

load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")
if openai_api_key:
    print("OPENAI_API_KEY is set")
else:
    print("OPENAI_API_KEY is not set")

# uv add python-dotenv ipykernel
```

```python
#Interact with LLM — response API
from httpx import Response
from openai import OpenAI
client = OpenAI()
response = client.responses.create(
    model="gpt-6-luna",
    input="Give three bullet points baout why Agentic AI is important?"
)
```

```python
#print(response)
print(response.output_text)
```

```python
# Method 1 for displaying the output text
from IPython.display import Markdown, display
display(Markdown(response.output_text))
```

```python
#alternative
messages = [
    {"role": "user", "content": "what is the capital of France?"}
]

client = OpenAI()

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=messages # input from previus line
)
print(response)

print(response.choices[0].message.content)
```
